```python
import numpy as np
import pandas as pd
import open3d as o3d
from plyfile import PlyData, PlyElement
import pyvista as pv
import os



# ============================================================
# 1. Compute native tree height
# ============================================================
def get_tree_height(wood_txt, foliage_txt):

    wood = pd.read_csv(
        wood_txt,
        sep=r"\s+",
        header=None
    )

    foliage = pd.read_csv(
        foliage_txt,
        sep=r"\s+",
        header=None
    )

    tree_height = max(
        wood[2].max(),
        foliage[2].max()
    )

    return tree_height



# ============================================================
# 2. Voxelization from TXT
# ============================================================
def build_voxels(txt_path, voxel_size):

    points = pd.read_csv(
        txt_path,
        sep=r"\s+",
        header=None
    )


    # XYZ coordinates
    xyz = points.iloc[:, :3].values


    # Use 4th column if available
    if points.shape[1] > 3:
        intensity = points.iloc[:, 3].astype(float).values
    else:
        intensity = np.ones(len(points))


    # voxel origin
    origin = xyz.min(axis=0)


    # voxel indices
    idx = np.floor(
        (xyz - origin) / voxel_size
    ).astype(int)



    df = pd.DataFrame(
        {
            "i": idx[:,0],
            "j": idx[:,1],
            "k": idx[:,2],
            "intensity": intensity
        }
    )


    voxel_volume = voxel_size ** 3



    vox = (
        df
        .groupby(["i","j","k"])
        .agg(
            point_count=("intensity","size"),
            mean_intensity=("intensity","mean")
        )
        .reset_index()
    )


    vox["point_density"] = (
        vox["point_count"] / voxel_volume
    )


    # voxel centre coordinates

    vox["x"] = (
        origin[0] +
        (vox["i"] + 0.5) * voxel_size
    )

    vox["y"] = (
        origin[1] +
        (vox["j"] + 0.5) * voxel_size
    )

    vox["z"] = (
        origin[2] +
        (vox["k"] + 0.5) * voxel_size
    )


    return vox




# ============================================================
# 3. Export voxel point cloud PLY
# ============================================================
def export_voxel_ply(vox_df, output):

    vertices = np.array(
        [
            (
                r.x,
                r.y,
                r.z,
                r.mean_intensity,
                r.point_density,
                r.point_count
            )

            for r in vox_df.itertuples()
        ],

        dtype=[
            ("x","f4"),
            ("y","f4"),
            ("z","f4"),
            ("intensity","f4"),
            ("density","f4"),
            ("count","i4")
        ]
    )


    PlyData(
        [
            PlyElement.describe(
                vertices,
                "vertex"
            )
        ]
    ).write(output)




# ============================================================
# 4. Export voxel cube mesh
# ============================================================
def export_voxel_mesh(
        vox_df,
        voxel_size,
        output
):

    mesh = o3d.geometry.TriangleMesh()


    max_density = vox_df.point_density.max()

    if max_density == 0:
        max_density = 1



    for row in vox_df.itertuples():

        cube = o3d.geometry.TriangleMesh.create_box(
            voxel_size,
            voxel_size,
            voxel_size
        )


        cube.translate(
            [
                row.x - voxel_size/2,
                row.y - voxel_size/2,
                row.z - voxel_size/2
            ]
        )


        val = row.point_density / max_density


        cube.paint_uniform_color(
            [
                val,
                0,
                1-val
            ]
        )


        mesh += cube



    mesh.remove_duplicated_vertices()
    mesh.remove_degenerate_triangles()
    mesh.compute_vertex_normals()


    o3d.io.write_triangle_mesh(
        output,
        mesh
    )




# ============================================================
# 5. Visualization
# ============================================================
def visualize(
        vox_df,
        voxel_size,
        output
):

    points = vox_df[
        ["x","y","z"]
    ].values


    cloud = pv.PolyData(points)


    cloud["density"] = (
        vox_df.point_density.values
    )


    cube = pv.Cube(
        x_length=voxel_size,
        y_length=voxel_size,
        z_length=voxel_size
    )


    glyph = cloud.glyph(
        geom=cube,
        scale=False,
        orient=False
    )


    plotter = pv.Plotter(
        off_screen=True
    )


    plotter.add_mesh(
        glyph,
        scalars="density",
        cmap="viridis"
    )


    plotter.add_axes()
    plotter.show_grid()


    plotter.show(
        screenshot=output
    )



# ============================================================
# 6. Main processing function
# ============================================================
def process(
        txt_file,
        voxel_size,
        name,
        outdir
):

    # Save beside input file
    folder = os.path.dirname(txt_file)


    vox = build_voxels(
        txt_file,
        voxel_size
    )


    print(
        f"{name}: {len(vox)} voxels"
    )


    export_voxel_ply(
        vox,
        os.path.join(
            outdir,
            f"{name}_voxels.ply"
        )
    )


    export_voxel_mesh(
        vox,
        voxel_size,
        os.path.join(
            outdir,
            f"{name}_cubes.ply"
        )
    )


    visualize(
        vox,
        voxel_size,
        os.path.join(
            outdir,
            f"{name}_render.png"
        )
    )


    return vox




# ============================================================
# 7. Run in Jupyter
# ============================================================

VOXEL_SIZE = 0.25
folders = [
    "Acacia",
    "Acacia10.0",
    "Acacia11.14",
    "Acacia11.68",
    "Acacia11.7",
    "Acacia12.85",
    "Acacia6.2",
    "Acacia7.14",
    "Prosopis1.93",
    "Prosopis3.21",
    "Prosopis3.94",
    "Prosopis5.0"
]

for tree_folder in folders:
# Folder relative to notebook location
    folder = os.path.join("Data", tree_folder)
    outdir = os.path.join("Data", "Voxelized_trees_0.25m", tree_folder)
    wood_file = os.path.join(
        folder,
        "wood.txt"
    )
    
    
    foliage_file = os.path.join(
        folder,
        "foliage.txt"
    )
    
    
    
    # Calculate whole tree height
    tree_height = get_tree_height(
        wood_file,
        foliage_file
    )
    
    
    print(
        f"Native tree height: {tree_height:.2f} m"
    )
    
    
    
    # Process wood
    wood = process(
        wood_file,
        VOXEL_SIZE,
        "wood",
        outdir
    
    )
    
    
    
    # Process foliage
    foliage = process(
        foliage_file,
        VOXEL_SIZE,
        "foliage",
        outdir
    )
```
